# Outliers

```python
def find_outlier_IQR(data):
    q1 = data.quantile(0.25)
    q3 = data.quantile(0.75)
    IQR = q3-q1

    outliers = data[(data < q1 - 1.5*IQR) | (data > q3 + 1.5*IQR)]
    return outliers

outliers = find_outlier_IQR(df["ApplicantIncome"])
print("Length of ouliers: " , len(outliers))
print("maximum outlier value: ", outliers.max())
print("minimum outlier value: ", outliers.min())
```

```python
q1 = df["ApplicantIncome"].quantile(0.25)
q3 = df["ApplicantIncome"].quantile(0.75)
IQR = q3-q1

lower_bound = q1 - 1.5*IQR
upper_bound = q3 + 1.5*IQR

df_cleaned = df[(df["ApplicantIncome"] >= lower_bound)   & (df["ApplicantIncome"] <= upper_bound )]

df_cleaned.shape
```
